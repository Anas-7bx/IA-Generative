## Partie I — Classification discriminative
### 2. Prédire la classe d’une image

**Objectif :** apprendre à associer une image `x` à une classe `y`.

**Explication :** le classifieur reçoit les pixels d’une image et prédit le chiffre correspondant. Il apprend la relation entre l’image et son étiquette ; il ne crée pas de nouvelles images.

```python
class Classifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 128),
            nn.ReLU(),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        return self.net(x)

classifier = Classifier().to(device)
classifier
```

### Fonction de coût de la classification

**Objectif :** mesurer l’écart entre la classe prédite et l’étiquette correcte.

**Explication :** `CrossEntropyLoss` convient à une classification à plusieurs classes. L’étiquette est nécessaire pour calculer la perte et mettre à jour le classifieur.

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(classifier.parameters(), lr=1e-3)
```

```python
def train_classifier(model, loader, epochs=3):
    model.train()

    for epoch in range(epochs):
        total_loss = 0

        for x, y in loader:
            x, y = x.to(device), y.to(device)

            logits = model(x)
            loss = criterion(logits, y)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        print(
            f"Epoch {epoch+1}/{epochs} - "
            f"Loss = {total_loss/len(loader):.4f}"
        )

train_classifier(classifier, train_loader, epochs=3)
```

```python
def evaluate_classifier(model, loader):
    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)

            logits = model(x)
            pred = logits.argmax(dim=1)

            correct += (pred == y).sum().item()
            total += y.size(0)

    return correct / total

accuracy = evaluate_classifier(classifier, test_loader)
print(f"Accuracy = {100*accuracy:.2f}%")
```

### Évaluer le classifieur

**Objectif :** mesurer la proportion d’images correctement classées.

**Explication :** la cellule suivante utilise l’accuracy pour compter les prédictions correctes. Cette mesure complète le suivi de la fonction de coût.

```python
from torcheval.metrics import MulticlassAccuracy

@torch.no_grad()
def evaluate_classifier(model, loader):
    model.eval()
    device = next(model.parameters()).device

    metric = MulticlassAccuracy(device=device)

    for x, y in loader:
        x, y = x.to(device), y.to(device)

        logits = model(x)
        metric.update(logits, y)

    return metric.compute().item()


accuracy = evaluate_classifier(classifier, test_loader)
print(f"Accuracy = {100 * accuracy:.2f}%")
```

## Partie II — Autoencodeur (AE)
### 3. Encoder puis reconstruire une image

**Objectif :** reconstruire une image à partir d’une représentation compacte.

**Explication :** l’encodeur transforme l’image `x` en vecteur latent `z`, puis le décodeur produit une reconstruction. Le modèle compare celle-ci à l’image d’origine. Les étiquettes ne sont pas nécessaires.

```python
class Autoencoder(nn.Module):
    def __init__(self, latent_dim=2):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 128),
            nn.ReLU(),
            nn.Linear(128, 32),
            nn.ReLU(),
            nn.Linear(32, latent_dim)
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 128),
            nn.ReLU(),
            nn.Linear(128, 784),
            nn.Sigmoid()
        )

    def encode(self, x):
        return self.encoder(x)

    def decode(self, z):
        x_hat = self.decoder(z)
        return x_hat.view(-1, 1, 28, 28)

    def forward(self, x):
        z = self.encode(x)
        x_hat = self.decode(z)
        return x_hat

ae = Autoencoder(latent_dim=2).to(device)
ae
```

### 4. Calculer l’erreur de reconstruction

**Objectif :** quantifier la différence entre l’image originale et sa reconstruction.

**Explication :** `MSELoss` mesure les écarts entre pixels. L’optimiseur utilise cette perte pour ajuster l’autoencodeur.

```python
ae_criterion = nn.MSELoss()
ae_optimizer = torch.optim.Adam(ae.parameters(), lr=1e-3)
def train_autoencoder(model, loader, epochs=5):
    model.train()

    for epoch in range(epochs):
        total_loss = 0

        for x, _ in loader:
            x = x.to(device)

            x_hat = model(x)
            loss = ae_criterion(x_hat, x)

            ae_optimizer.zero_grad()
            loss.backward()
            ae_optimizer.step()

            total_loss += loss.item()

        print(
            f"Epoch {epoch+1}/{epochs} - "
            f"Loss = {total_loss/len(loader):.4f}"
        )

train_autoencoder(ae, train_loader, epochs=5)
```

### Sélectionner un chiffre connu

**Objectif :** afficher un exemple MNIST appartenant à la classe choisie.

**Explication :** la fonction recherche une image existante portant l’étiquette demandée. Il s’agit d’une sélection dans le jeu de données, et non d’une génération.

```python
import torch
import matplotlib.pyplot as plt

@torch.no_grad()
def generer_chiffre(model, dataloader, chiffre):
    if chiffre not in range(10):
        raise ValueError("Le chiffre doit être compris entre 0 et 9.")

    device = next(model.parameters()).device
    model.eval()

    somme_z = None
    nombre = 0

    for images, labels in dataloader:
        # Sélection des images du chiffre demandé
        images = images[labels == chiffre]

        if images.shape[0] == 0:
            continue

        z = model.encode(images.to(device))

        if somme_z is None:
            somme_z = z.sum(dim=0, keepdim=True)
        else:
            somme_z += z.sum(dim=0, keepdim=True)

        nombre += z.shape[0]

    if nombre == 0:
        raise ValueError(f"Aucune image du chiffre {chiffre} trouvée.")

    z_moyen = somme_z / nombre
    return model.decode(z_moyen).cpu()
```

### 7. Limite de l’autoencodeur classique

**Objectif :** comprendre pourquoi un AE n’est pas toujours adapté à la génération.

**Explication :** l’AE apprend à reconstruire les exemples, mais ne force pas les codes latents à suivre une distribution régulière connue. Un point choisi au hasard peut produire une image peu réaliste. Cette limite motive le VAE.

## Partie III — Autoencodeur variationnel (VAE)
### 8. Représenter une distribution latente

**Objectif :** apprendre un espace latent régulier dans lequel on peut échantillonner.

**Explication :** le VAE associe à l’image une distribution décrite par une moyenne `mu` et une variance. Un vecteur latent est tiré à partir de cette distribution.

### 9. Utiliser l’astuce de reparamétrisation

**Objectif :** tirer un vecteur latent tout en conservant le calcul des gradients.

**Explication :** on tire un bruit `epsilon`, puis on le combine avec la moyenne `mu` et l’écart-type `sigma` : `z = mu + sigma * epsilon`. Cette écriture permet la rétropropagation dans l’encodeur.

```python
class VAE(nn.Module):
    def __init__(self, latent_dim=2):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 256),
            nn.ReLU(),
            nn.Linear(256, 64),
            nn.ReLU()
        )

        self.fc_mu = nn.Linear(64, latent_dim)
        self.fc_logvar = nn.Linear(64, latent_dim)

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def encode(self, x):
        h = self.encoder(x)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + std * eps

    def decode(self, z):
        x_hat = self.decoder(z)
        return x_hat.view(-1, 1, 28, 28)

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        x_hat = self.decode(z)
        return x_hat, mu, logvar

vae = VAE(latent_dim=2).to(device)
vae
```

### 10. Fonction de coût et entraînement du VAE

**Objectif :** combiner la qualité des reconstructions et la régularité de l’espace latent.

**Explication :** la perte comprend l’erreur de reconstruction et une divergence KL qui rapproche la distribution latente d’une loi normale standard. Le paramètre `beta` règle le poids de cette régularisation.

```python
def vae_loss(x_hat, x, mu, logvar, beta=1.0):
    rec_loss = F.binary_cross_entropy(
        x_hat,
        x,
        reduction="sum"
    )

    kl_loss = -0.5 * torch.sum(
        1 + logvar - mu.pow(2) - logvar.exp()
    )

    return rec_loss + beta * kl_loss, rec_loss, kl_loss
```

```python
## Entrainement
vae_optimizer = torch.optim.Adam(vae.parameters(), lr=1e-3)
def train_vae(model, loader, epochs=20, beta=1.0):
    model.train()

    for epoch in range(epochs):
        total_loss = 0
        total_rec = 0
        total_kl = 0

        for x, _ in loader:
            x = x.to(device)

            x_hat, mu, logvar = model(x)

            loss, rec, kl = vae_loss(
                x_hat,
                x,
                mu,
                logvar,
                beta=beta
            )

            vae_optimizer.zero_grad()
            loss.backward()
            vae_optimizer.step()

            total_loss += loss.item()
            total_rec += rec.item()
            total_kl += kl.item()

        n = len(loader.dataset)

        print(
            f"Epoch {epoch+1}/{epochs} | "
            f"Loss={total_loss/n:.2f} | "
            f"Rec={total_rec/n:.2f} | "
            f"KL={total_kl/n:.2f}"
        )

train_vae(vae, train_loader, epochs=20)
```

## Partie IV — Réseaux antagonistes génératifs (GAN)
### 15. Comprendre le principe des GAN

**Objectif :** comprendre le rôle du générateur et du discriminateur.

**Explication :** le générateur transforme un bruit aléatoire en image synthétique. Le discriminateur distingue les images réelles des images générées. Le générateur cherche à tromper le discriminateur, qui apprend à reconnaître les fausses images.

### Mise en place d’un GAN sur MNIST

**Objectif :** entraîner un GAN entièrement connecté à produire des chiffres manuscrits.

**Explication :** le générateur crée une image à partir d’un vecteur aléatoire, tandis que le discriminateur évalue si elle semble réelle. Ce GAN n’est pas conditionnel et ne reçoit pas la classe à générer.

### 1. Préparer les données du GAN

**Objectif :** charger MNIST et adapter l’intervalle des pixels à la sortie du générateur.

**Explication :** `ToTensor()` transforme les images en tenseurs, puis `Normalize` ramène les pixels entre -1 et 1, comme la fonction `Tanh`. Le `DataLoader` regroupe et mélange les exemples. Les étiquettes ne sont pas utilisées.

```python
from pathlib import Path

import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from torchvision.utils import save_image

# Facilite la reproductibilité, sans la garantir sur tous les matériels.
torch.manual_seed(42)

# Utiliser le GPU CUDA s'il est disponible.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

latent_dim = 100
image_dim = 28 * 28
batch_size = 128
epochs = 20
learning_rate = 0.0002

output_dir = Path("images_generees")
output_dir.mkdir(exist_ok=True)

print(f"Appareil utilisé : {device}")
```

### Paramètres de l’expérience GAN

**Objectif :** définir les dimensions et les réglages de l’entraînement.

**Explication :** `latent_dim` définit la taille du bruit, `image_dim` le nombre de pixels, `batch_size` la taille des lots, `epochs` le nombre de parcours et `learning_rate` le pas de mise à jour.

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,)),
])

dataset = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=transform,
)

loader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=True,
)
```

### 2. Construire le générateur

**Objectif :** transformer un vecteur de bruit en image MNIST.

**Explication :** le réseau transforme le vecteur latent jusqu’aux 784 pixels de l’image. `LeakyReLU` apporte une non-linéarité et `Tanh` borne les valeurs entre -1 et 1. La sortie est remise en forme d’image.

```python
class Generator(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.LeakyReLU(0.2),

            nn.Linear(256, 512),
            nn.LeakyReLU(0.2),

            nn.Linear(512, image_dim),
            nn.Tanh(),
        )

    def forward(self, noise):
        return self.network(noise)


generator = Generator(latent_dim).to(device)
```

### 3. Construire le discriminateur

**Objectif :** attribuer un score à une image pour distinguer les exemples réels des exemples générés.

**Explication :** l’image est aplatie en 784 valeurs puis traitée par des couches entièrement connectées. La sortie est un logit. `BCEWithLogitsLoss` inclut déjà la sigmoid ; il ne faut pas en ajouter une autre.

```python
class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(image_dim, 512),
            nn.LeakyReLU(0.2),

            nn.Linear(512, 256),
            nn.LeakyReLU(0.2),

            nn.Linear(256, 1),
        )

    def forward(self, images):
        return self.network(images)


discriminator = Discriminator().to(device)
```

### 4. Préparer les optimiseurs

**Objectif :** définir la mise à jour séparée des deux réseaux.

**Explication :** le générateur et le discriminateur ont chacun leur optimiseur Adam. Un vecteur de bruit fixe est conservé pour comparer les images produites à chaque époque.

```python
criterion = nn.BCEWithLogitsLoss()

optimizer_g = torch.optim.Adam(
    generator.parameters(),
    lr=learning_rate,
    betas=(0.5, 0.999),
)

optimizer_d = torch.optim.Adam(
    discriminator.parameters(),
    lr=learning_rate,
    betas=(0.5, 0.999),
)
```

```python
# Même bruit à chaque époque pour comparer les résultats.
fixed_noise = torch.randn(64, latent_dim, device=device)

for epoch in range(epochs):
    generator.train()
    discriminator.train()

    total_loss_d = 0.0
    total_loss_g = 0.0

    for real_images, _ in loader:
        # [B, 1, 28, 28] -> [B, 784]
        real_images = real_images.to(device).flatten(start_dim=1)

        # Le dernier lot peut contenir moins de 128 images.
        current_batch_size = real_images.size(0)

        real_labels = torch.ones(
            current_batch_size, 1, device=device
        )
        fake_labels = torch.zeros(
            current_batch_size, 1, device=device
        )

        # =====================================
        # A. Entraîner le discriminateur
        # =====================================
        discriminator.requires_grad_(True)
        optimizer_d.zero_grad(set_to_none=True)

        # Évaluer les vraies images.
        real_scores = discriminator(real_images)
        loss_real = criterion(real_scores, real_labels)

        # Créer de fausses images à partir de bruit.
        noise = torch.randn(
            current_batch_size, latent_dim, device=device
        )
        fake_images = generator(noise)

        # Couper le gradient vers G pendant cette étape.
        fake_scores = discriminator(fake_images.detach())
        loss_fake = criterion(fake_scores, fake_labels)

        loss_d = (loss_real + loss_fake) / 2

        loss_d.backward()
        optimizer_d.step()

        # =====================================
        # B. Entraîner le générateur
        # =====================================
        discriminator.requires_grad_(False)
        optimizer_g.zero_grad(set_to_none=True)

        # Nouvelle évaluation avec D qui vient d'être mis à jour.
        fake_scores = discriminator(fake_images)

        # G veut que ses images soient classées "réelles".
        loss_g = criterion(fake_scores, real_labels)

        loss_g.backward()
        optimizer_g.step()

        total_loss_d += loss_d.item()
        total_loss_g += loss_g.item()

    print(
        f"Époque {epoch + 1:02d}/{epochs} | "
        f"Loss D : {total_loss_d / len(loader):.4f} | "
        f"Loss G : {total_loss_g / len(loader):.4f}"
    )

    # =====================================
    # C. Sauvegarder des exemples
    # =====================================
    generator.eval()

    with torch.no_grad():
        samples = generator(fixed_noise)
        samples = samples.view(-1, 1, 28, 28)

        # [-1, 1] -> [0, 1] pour enregistrer les images.
        samples = (samples + 1) / 2

        save_image(
            samples,
            output_dir / f"epoch_{epoch + 1:03d}.png",
            nrow=8,
        )

# Sauvegarder les paramètres appris.
torch.save(generator.state_dict(), "generator_mnist.pt")
```

### Implémentation complète du GAN

**Objectif :** rassembler les composants de l’expérience GAN dans une cellule.

**Explication :** la cellule suivante regroupe le chargement des données, les deux réseaux et leur entraînement. Elle récapitule le code présenté plus haut.

```python
from pathlib import Path

import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from torchvision.utils import save_image

# -------------------------
# Configuration
# -------------------------
torch.manual_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

latent_dim = 100
batch_size = 128
epochs = 20
learning_rate = 2e-4

output_dir = Path("images_generees")
output_dir.mkdir(exist_ok=True)

# -------------------------
# Données : images de 28 × 28
# -------------------------
transform = transforms.Compose([
    transforms.ToTensor(),
    # Passage de [0, 1] à [-1, 1], comme la sortie du générateur
    transforms.Normalize((0.5,), (0.5,)),
])

dataset = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=transform,
)

loader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=True,
)

# -------------------------
# Générateur : bruit -> image
# -------------------------
generator = nn.Sequential(
    nn.Linear(latent_dim, 256),
    nn.LeakyReLU(0.2),

    nn.Linear(256, 512),
    nn.LeakyReLU(0.2),

    nn.Linear(512, 28 * 28),
    nn.Tanh(),  # Pixels dans [-1, 1]
).to(device)

# -------------------------
# Discriminateur : image -> score
# -------------------------
discriminator = nn.Sequential(
    nn.Linear(28 * 28, 512),
    nn.LeakyReLU(0.2),

    nn.Linear(512, 256),
    nn.LeakyReLU(0.2),

    nn.Linear(256, 1),
    # Pas de Sigmoid : inclus dans BCEWithLogitsLoss
).to(device)

criterion = nn.BCEWithLogitsLoss()

optimizer_g = torch.optim.Adam(
    generator.parameters(),
    lr=learning_rate,
    betas=(0.5, 0.999),
)

optimizer_d = torch.optim.Adam(
    discriminator.parameters(),
    lr=learning_rate,
    betas=(0.5, 0.999),
)

# Même bruit à chaque époque pour visualiser la progression
fixed_noise = torch.randn(64, latent_dim, device=device)

# -------------------------
# Entraînement
# -------------------------
for epoch in range(epochs):
    total_loss_d = 0.0
    total_loss_g = 0.0

    for real_images, _ in loader:
        real_images = real_images.to(device).view(-1, 28 * 28)
        current_batch_size = real_images.size(0)

        real_labels = torch.ones(current_batch_size, 1, device=device)
        fake_labels = torch.zeros(current_batch_size, 1, device=device)

        # 1. Entraîner le discriminateur
        discriminator.requires_grad_(True)
        optimizer_d.zero_grad(set_to_none=True)

        real_scores = discriminator(real_images)
        loss_real = criterion(real_scores, real_labels)

        noise = torch.randn(current_batch_size, latent_dim, device=device)
        fake_images = generator(noise)

        # detach() empêche de modifier le générateur pendant cette étape
        fake_scores = discriminator(fake_images.detach())
        loss_fake = criterion(fake_scores, fake_labels)

        loss_d = (loss_real + loss_fake) / 2
        loss_d.backward()
        optimizer_d.step()

        # 2. Entraîner le générateur
        # On fige les paramètres de D, mais on garde le gradient vers G.
        discriminator.requires_grad_(False)
        optimizer_g.zero_grad(set_to_none=True)

        fake_scores = discriminator(fake_images)

        # G cherche à faire classer ses images comme "vraies".
        loss_g = criterion(fake_scores, real_labels)

        loss_g.backward()
        optimizer_g.step()

        total_loss_d += loss_d.item()
        total_loss_g += loss_g.item()

    print(
        f"Époque {epoch + 1:02d}/{epochs} | "
        f"Loss D : {total_loss_d / len(loader):.4f} | "
        f"Loss G : {total_loss_g / len(loader):.4f}"
    )

    # Sauvegarder une grille de 64 images
    generator.eval()
    with torch.no_grad():
        samples = generator(fixed_noise).view(-1, 1, 28, 28)

        # Repasser de [-1, 1] à [0, 1]
        samples = (samples + 1) / 2

        save_image(
            samples,
            output_dir / f"epoch_{epoch + 1:03d}.png",
            nrow=8,
        )
    generator.train()
# Mode évaluation
generator.eval()

# Générer 16 images sans calculer de gradients
with torch.no_grad():
    noise = torch.randn(16, latent_dim, device=device)
    images = generator(noise)

    # [16, 784] -> [16, 1, 28, 28]
    images = images.reshape(-1, 1, 28, 28)

    # Convertir les pixels de [-1, 1] vers [0, 1]
    images = ((images + 1) / 2).clamp(0, 1)

    # Matplotlib utilise les données sur CPU
    images = images.cpu()

# Afficher une grille de 4 × 4
fig, axes = plt.subplots(4, 4, figsize=(6, 6))

for ax, image in zip(axes.flat, images):
    ax.imshow(image.squeeze(0).numpy(), cmap="gray", vmin=0, vmax=1)
    ax.axis("off")

fig.suptitle("Chiffres générés par le GAN")
plt.tight_layout()
plt.show()
# Sauvegarder les poids du générateur
torch.save(generator.state_dict(), "generator_mnist.pt")
```

## Partie V — Modèles de diffusion

**Objectif :** étudier une méthode de génération fondée sur l’apprentissage du débruitage.

### Comprendre le principe de la diffusion

**Objectif :** décrire comment un modèle de diffusion peut générer une image.

**Explication :** dans le processus direct, du bruit est progressivement ajouté à une image réelle. Le réseau apprend à prédire le bruit. Pour générer, on part d’un bruit aléatoire et le modèle le retire progressivement.

### 1. Préparer les données et le périphérique

**Objectif :** charger MNIST, transformer les images et choisir le support de calcul.

**Explication :** les images sont converties en tenseurs puis normalisées entre -1 et 1. Le `DataLoader` les regroupe en lots. Le modèle apprend ici à générer sans utiliser les étiquettes.

```python
import math
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt

from torch.utils.data import DataLoader
from torchvision import datasets, transforms

torch.manual_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Périphérique :", device)

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

train_loader = DataLoader(
    train_dataset,
    batch_size=128,
    shuffle=True,
    num_workers=0,
    pin_memory=(device.type == "cuda")
)
```

### 2. Définir le calendrier d’ajout de bruit

**Objectif :** contrôler la quantité de bruit appliquée à chaque instant `t`.

**Explication :** le calendrier calcule `beta`, `alpha` et leur produit cumulé `alpha_bar`. La fonction combine l’image initiale `x0` avec un bruit gaussien `epsilon` pour obtenir l’image bruitée `xt`. Les coefficients sont mis en forme pour correspondre aux dimensions du lot.

```python
T = 200

def cosine_beta_schedule(timesteps, s=0.008):
    steps = torch.linspace(0, timesteps, timesteps + 1)

    alpha_bar = torch.cos(
        ((steps / timesteps + s) / (1 + s)) * math.pi / 2
    ).pow(2)

    alpha_bar = alpha_bar / alpha_bar[0]

    betas = 1 - alpha_bar[1:] / alpha_bar[:-1]

    return betas.clamp(min=1e-4, max=0.999)


betas = cosine_beta_schedule(T).to(device)
alphas = 1.0 - betas
alpha_bars = torch.cumprod(alphas, dim=0)

alpha_bars_prev = torch.cat([
    torch.ones(1, device=device),
    alpha_bars[:-1]
])

# Variance utilisée pendant la génération.
posterior_variance = (
    betas * (1 - alpha_bars_prev) / (1 - alpha_bars)
)


def extract(values, t):
    """Sélectionner un coefficient par image et permettre le broadcasting."""
    return values[t].view(-1, 1, 1, 1)


def add_noise(x0, t, noise=None):
    if noise is None:
        noise = torch.randn_like(x0)

    signal = extract(alpha_bars.sqrt(), t)
    noise_scale = extract((1 - alpha_bars).sqrt(), t)

    xt = signal * x0 + noise_scale * noise

    return xt, noise
```

### 3. Observer le bruitage progressif d’une image

**Objectif :** visualiser la dégradation d’un chiffre quand le niveau de bruit augmente.

**Explication :** une image MNIST est bruitée à plusieurs instants puis affichée sur la même échelle. Le bruit est réutilisé pour faciliter la comparaison. Cette cellule illustre le processus direct, pas la génération inverse.

```python
x0, _ = train_dataset[0]
x0 = x0.unsqueeze(0).to(device)

times = [0, 40, 80, 120, 160, 199]
noise = torch.randn_like(x0)

fig, axes = plt.subplots(1, len(times) + 1, figsize=(14, 2))

axes[0].imshow(
    x0[0, 0].cpu(),
    cmap="gray",
    vmin=-1,
    vmax=1
)
axes[0].set_title("Original")
axes[0].axis("off")

for ax, step in zip(axes[1:], times):
    t = torch.tensor([step], device=device)
    xt, _ = add_noise(x0, t, noise)

    ax.imshow(
        xt[0, 0].cpu(),
        cmap="gray",
        vmin=-1,
        vmax=1
    )
    ax.set_title(f"t = {step}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

### 4. Construire le réseau qui prédit le bruit

**Objectif :** définir un réseau qui estime le bruit présent dans une image bruitée.

**Explication :** le modèle reçoit l’image et l’instant `t`. Un encodage temporel indique le niveau de bruit ; les blocs convolutifs traitent l’image en intégrant cette information. La sortie a la même forme que le bruit à prédire.

```python
class TimeEmbedding(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, t):
        half = self.dim // 2

        frequencies = torch.exp(
            -math.log(10000)
            * torch.arange(
                half,
                device=t.device,
                dtype=torch.float32
            )
            / (half - 1)
        )

        angles = t.float()[:, None] * frequencies[None, :]

        return torch.cat([
            angles.sin(),
            angles.cos()
        ], dim=1)


class ConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_dim):
        super().__init__()

        self.conv1 = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, 3, padding=1),
            nn.GroupNorm(8, out_channels),
            nn.SiLU()
        )

        self.time_projection = nn.Linear(time_dim, out_channels)

        self.conv2 = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, 3, padding=1),
            nn.GroupNorm(8, out_channels),
            nn.SiLU()
        )

    def forward(self, x, time_embedding):
        h = self.conv1(x)

        # Une information temporelle par canal.
        time_features = self.time_projection(time_embedding)
        h = h + time_features[:, :, None, None]

        return self.conv2(h)


class SmallUNet(nn.Module):
    def __init__(self, time_dim=128):
        super().__init__()

        self.time_embedding = nn.Sequential(
            TimeEmbedding(time_dim),
            nn.Linear(time_dim, time_dim),
            nn.SiLU(),
            nn.Linear(time_dim, time_dim)
        )

        self.encoder1 = ConvBlock(1, 32, time_dim)
        self.encoder2 = ConvBlock(32, 64, time_dim)
        self.middle = ConvBlock(64, 128, time_dim)

        self.pool = nn.AvgPool2d(2)

        self.decoder2 = ConvBlock(128 + 64, 64, time_dim)
        self.decoder1 = ConvBlock(64 + 32, 32, time_dim)

        self.output = nn.Conv2d(32, 1, kernel_size=1)

    def forward(self, x, t):
        time_embedding = self.time_embedding(t)

        h1 = self.encoder1(x, time_embedding)               # 28 × 28
        h2 = self.encoder2(self.pool(h1), time_embedding)   # 14 × 14
        h = self.middle(self.pool(h2), time_embedding)     # 7 × 7

        h = F.interpolate(
            h, size=h2.shape[-2:], mode="nearest"
        )
        h = self.decoder2(
            torch.cat([h, h2], dim=1),
            time_embedding
        )

        h = F.interpolate(
            h, size=h1.shape[-2:], mode="nearest"
        )
        h = self.decoder1(
            torch.cat([h, h1], dim=1),
            time_embedding
        )

        return self.output(h)


diffusion_model = SmallUNet().to(device)
```

### 5. Entraîner le modèle à prédire le bruit

**Objectif :** apprendre au réseau à retrouver le bruit ajouté aux images.

**Explication :** pour chaque lot, un instant et un bruit sont tirés. L’image est bruitée, puis le modèle prédit le bruit. La MSE compare la prédiction au bruit utilisé et sert à ajuster le réseau.

```python
optimizer = torch.optim.Adam(
    diffusion_model.parameters(),
    lr=2e-4
)


def train_diffusion(model, loader, optimizer, epochs=20):
    history = []

    for epoch in range(epochs):
        model.train()

        total_loss = 0.0
        n_images = 0

        for x, _ in loader:
            x = x.to(device)
            batch_size = x.size(0)

            # Un niveau de bruit différent pour chaque image.
            t = torch.randint(
                0, T,
                (batch_size,),
                device=device
            )

            xt, true_noise = add_noise(x, t)
            predicted_noise = model(xt, t)

            loss = F.mse_loss(predicted_noise, true_noise)

            optimizer.zero_grad()
            loss.backward()

            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

            total_loss += loss.item() * batch_size
            n_images += batch_size

        mean_loss = total_loss / n_images
        history.append(mean_loss)

        print(
            f"Epoch {epoch + 1}/{epochs} | "
            f"MSE bruit = {mean_loss:.4f}"
        )

    return history


history = train_diffusion(
    diffusion_model,
    train_loader,
    optimizer,
    epochs=20
)
```

### 6. Générer des images par diffusion inverse

**Objectif :** transformer progressivement un bruit aléatoire en images.

**Explication :** la fonction part d’un tenseur bruité et parcourt les instants dans l’ordre inverse. À chaque étape, la prédiction du réseau permet d’actualiser l’image. Les gradients ne sont pas calculés, car la fonction sert à générer et non à entraîner.

```python
@torch.no_grad()
def generate_images(model, n_samples=20):
    model.eval()
    model_device = next(model.parameters()).device

    # Réutiliser exactement le calendrier de l'entraînement.
    b = betas.to(model_device)
    a = 1.0 - b
    abar = torch.cumprod(a, dim=0)

    abar_prev = torch.cat([
        torch.ones(1, device=model_device),
        abar[:-1]
    ])

    # Variance du processus inverse.
    posterior_var = (
        b * (1.0 - abar_prev) / (1.0 - abar)
    )

    # Coefficients de la moyenne conditionnelle q(x_{t-1} | x_t, x_0).
    coef_x0 = (
        b * torch.sqrt(abar_prev) / (1.0 - abar)
    )

    coef_xt = (
        torch.sqrt(a) * (1.0 - abar_prev) / (1.0 - abar)
    )

    x = torch.randn(
        n_samples, 1, 28, 28,
        device=model_device
    )

    for step in reversed(range(len(b))):
        t = torch.full(
            (n_samples,),
            step,
            device=model_device,
            dtype=torch.long
        )

        predicted_noise = model(x, t)

        # Estimation de l'image propre.
        x0_pred = (
            x - torch.sqrt(1.0 - abar[step]) * predicted_noise
        ) / torch.sqrt(abar[step])

        # Les images propres d'entraînement sont dans [-1, 1].
        x0_pred = x0_pred.clamp(-1.0, 1.0)

        # Moyenne de l'étape précédente.
        mean = (
            coef_x0[step] * x0_pred
            + coef_xt[step] * x
        )

        if step > 0:
            x = (
                mean
                + torch.sqrt(posterior_var[step])
                * torch.randn_like(x)
            )
        else:
            x = mean

        if not torch.isfinite(x).all():
            raise RuntimeError(
                f"Valeurs non finies pendant la génération, étape {step}"
            )

    return ((x + 1.0) / 2.0).clamp(0, 1).cpu()
```
